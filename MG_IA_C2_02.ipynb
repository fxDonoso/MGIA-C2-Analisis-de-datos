# Grafico de tendencia

```python
from matplotlib.lines import lineStyles
plt.figure(figsize=(10,4))
plt.plot(df["INDICE_OCUPACIONAL"])
plt.axhline(df["INDICE_OCUPACIONAL"].mean(),linestyle="--", color="red")
plt.title("Tendencias de índice ocupacional")
plt.ylabel("Índice ocupacional %")
plt.xlabel("Observaciones")
plt.show()
```
