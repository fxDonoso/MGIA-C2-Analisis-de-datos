## Revisión de duplicados

```python
# Total de filas duplicadas (todas las columnas)
n_duplicados = df.duplicated().sum()
print(f"Filas completamente duplicadas: {n_duplicados}")

# Mostrar las filas duplicadas si existen
duplicados = df[df.duplicated(keep=False)]
if not duplicados.empty:
    print(f"\nRegistros duplicados ({len(duplicados)} filas):")
    display(duplicados)
else:
    print("No hay filas completamente duplicadas.")

# Revisar duplicados por columnas clave de identidad
cols_clave = ["RUT", "Nombre de usuario", "Dirección de correo"]
for col in cols_clave:
    dupes = df[df.duplicated(subset=[col], keep=False) & df[col].notna()]
    print(f"\nDuplicados por '{col}': {df.duplicated(subset=[col]).sum()}")
    if not dupes.empty:
        display(dupes[[col, "Nombre", "Apellido(s)", "Institución"]].sort_values(col, key=lambda x: x.astype(str)))
```
