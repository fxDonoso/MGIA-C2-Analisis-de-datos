```python
valor = 250000
lista_clientes = [{
    "nombre" : "Maria Gonzalez",
    "empresa" : "Innova Ltda",
    "compras" : 150
},{
    "nombre" : "Mario Diaz",
    "empresa" : "JZP Ltda",
    "compras" : 10
},{
    "nombre" : "Juanito Zabala",
    "empresa" : "Coproin Ltda",
    "compras" : 1500
},{
    "nombre" : "Joaquin l. Martin",
    "empresa" : "Uranus Ltda",
    "compras" : 165
}]
```

```python
def calcular_compra(cantidad):
  resultado = cantidad * valor
  return resultado
```

```python
for cliente in lista_clientes:
  if cliente["compras"] > 100:
    print(f"- nombre cliente: {cliente["nombre"]} cantidad de compras {cliente["compras"]} y el total es de ${calcular_compra(cliente["compras"])}")
```

```python
for cliente in lista_clientes:
  print(f"Cliente: {cliente['nombre']}")
  print(f"Empresa: {cliente['empresa']}")
  print(f"Compras: {cliente['compras']}\n")
```

```python
def agregar_cliente(nombre, empresa, compras):
  cliente = {
      "nombre" : nombre,
      "empresa" : empresa,
      "compras" : compras
  }
  lista_clientes.append(cliente)
```

```python
agregar_cliente("Hector", "National CDI", 56)
```

```python
print(lista_clientes)
```

```python
def eliminar_cliente(nombre):
  for cliente in lista_clientes:
    if cliente["nombre"] == nombre:
      lista_clientes.remove(cliente)
```

```python
def actualizar_cliente(nombre, empresa, compras):
  for cliente in lista_clientes:
    if cliente["nombre"] == nombre:
      cliente["empresa"] = empresa
      cliente["compras"] = compras
```

```python
def visualizar_cliente(nombre):
  for cliente in lista_clientes:
    if cliente["nombre"] == nombre:
      print(f"Cliente: {cliente['nombre']}")
      print(f"Empresa: {cliente['empresa']}")
      print(f"Compras: {cliente['compras']}\n")

```

```python
visualizar_cliente("Juanito Zabala")
```

Gráfico de tendencia (índice ocupacional)

```python
plt.figure(figsize=(10,4))
plt.plot(df["INDICE_OCUPACIONAL"])
plt.axhline(df["INDICE_OCUPACIONAL"].mean(), linestyle="--")

plt.title("Tendencia del índice ocupacional")
plt.ylabel("Indice Ocupacional (%)")
plt.xlabel("Observaciones")
plt.show()
```
