```python


lista_notas = []
while(True):
  print("------------------------------")
  print(" == MENU NOTAS ==")
  print("1) Agregar nota")
  print("2) Modificar nota")
  print("3) Eliminar nota")
  print("4) Mostras notas")
  print("5) Promedio notas")
  print("6) Salir")

  opcion = input("Ingrese una opcion: ")

  if opcion == "1":
    nota = float(input("Ingrese una calificación: "))
    lista_notas.append(nota)
    print("La nota ", nota, " ha sido agregada.")

  elif opcion == "2":
    print("Ha seleccionado modificar nota")
    print("Las notas actuales son: ")
    for i in range(len(lista_notas)):
      print(f"Posición {i} -> Nota: {lista_notas[i]}")
    print("---")
    nota_modificar = float(input("Ingrese la nota a modificar: "))
    posicion_modificar = int(input("Ingrese la posición: "))
    lista_notas[posicion_modificar] = nota_modificar
    print("La nota ha sido modificada.")
  elif opcion == "3":
    print("Ha seleccionado eliminar una nota")
    for i in range(len(lista_notas)):
      print(f"Posición {i} -> Nota: {lista_notas[i]}")
    print("---")
    nota_eliminar = int(input("Ingrese el índice de la nota a eliminar: "))
    del lista_notas[nota_eliminar]
    print("La nota ha sido eliminada.")
  elif opcion == "4":
    print("Mostrar notas:")
    for nota in lista_notas:
      print("Nota: ", nota)
    print("-----------")

  elif opcion == "5":
    print("Ha seleccionado calcular promedio: ")
    print(f"El promedio de las notas actuales es: {sum(lista_notas)/len(lista_notas)}")
    print("---------------")
  elif opcion == "6":
    print("Ha seleccionado salir.")
    break
  else:
    print("Opcion incorrecta.")
```

```python
inventario = {}

def agregar_equipo():
  codigo = input("Ingresar el código del equipo: ")
  if codigo in inventario:
    print("El equipo ya existe.")
  else:
    marca = input("Ingresar marca: ")
    modelo = input("Ingresar modelo: ")
    cantidad = int(input("Ingrese cantidad: "))

    inventario[codigo] = {
        "marca" : marca,
        "modelo" : modelo,
        "cantidad" : cantidad
    }

    print("Equipo agregado correctamente.")

def mostrar_inventario():

  if len(inventario) == 0:
    print("No hay equipos registrados.")
  else:
    print("---------EQUIPOS EN INVENTARIO ---------")

    for codigo, datos in inventario.items():
      print("Codigo: ", codigo)
      print("Marca: ", datos["marca"])
      print("Modelo: ", datos["modelo"])
      print("Cantidado: ", datos["cantidad"])
      print("-----------")

def modificar_equipo():

  codigo = input("Ingresar el código del equipo: ")

  if codigo in inventario:
    nueva_marca = input("Nueva marca: ")
    nuevo_modelo = input("Nueva modelo: ")
    nueva_cantidad = input("Nueva cantidad: ")

    inventario[codigo]["marca"] = nueva_marca
    inventario[codigo]["modelo"] = nuevo_modelo
    inventario[codigo]["cantidad"] = nueva_cantidad

    print("Se ha modificado correctamente.")
  else:
    print("Equipo no encontrado.")

def eliminar_equipo():
  codigo = input("Ingresar el código del equipo: ")

  if codigo in inventario:
    del inventario[codigo]
    print("Equipo eliminado correctamente.")
  else:
    print("Equipo no encontrado.")

while(True):

  print(" ==== SISTEMA DE INVENTARIO ====")
  print("1) Agregar equipo")
  print("2) Mostrar inventario")
  print("3) Modificar equipo")
  print("4) Eliminar equipo")
  print("5) Salir")

  opcion = input("Ingrese opcion: ")

  if opcion == "1":
    agregar_equipo()
  elif opcion == "2":
    mostrar_inventario()
  elif opcion == "3":
    modificar_equipo()
  elif opcion == "4":
    eliminar_equipo()
  elif opcion == "5":
    print("Ha salido.")
    break
  else:
    print("Opcion inválida.")
```
