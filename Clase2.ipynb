# Funciones

```python

valor=250000
lista_clientes= [{
    "nombre": "Juan",
    "empresa": "Tech Solutions",
    "compras": 5
},{
    "nombre": "Maria",
    "empresa": "Innovatech",
    "compras": 3
},{
    "nombre": "Pedro",
    "empresa": "DataCorp",
    "compras": 7
},{
    "nombre": "Ana",
    "empresa": "SoftWorks",
    "compras": 2
}]
```

```python
def calcular_compra(cantidad):
    resultado = cantidad * valor
    return resultado
```

```python
for cliente in lista_clientes:
    print(f"El cliente {cliente['nombre']} de la empresa {cliente['empresa']} ha realizado {cliente['compras']} compras.")
    total_compra = calcular_compra(cliente['compras'])
    print(f"El total de la compra es: ${total_compra}") 
```

## Ejercicio 1
 * Crear lista con tiempos de atención.
* Calcular promedio.
* Usar condicional para emitir alerta si supera cierto límite.

```python
tiempos_de_atencion = [15, 20, 10, 25]
promedio_atencion = sum(tiempos_de_atencion) / len(tiempos_de_atencion)
print(f"El tiempo promedio de atención al cliente es: {promedio_atencion} minutos")
if promedio_atencion > 15:
    print("El tiempo de atención es alto, se recomienda mejorar la eficiencia.")
```
