## Transformar valores a Numericos

```python
df["tiempo_entrega_horas"] = pd.to_numeric(df["tiempo_entrega_horas"], errors="coerce")
df
```

## Medidas de Tendencia Central

```python
media = df["tiempo_entrega_horas"].mean()
print("La media de tiempos de entrega es: ", media)
```

```python
mediana = df["tiempo_entrega_horas"].median()
print("La mediana es: ", mediana)
```

```python
moda = df["tiempo_entrega_horas"].mode()
print("La moda es: ", moda)
```

## Medidas de Disperción

### Varianza

```python
varianza = df["tiempo_entrega_horas"].var()
varianza
```

### Desviación estandar

```python
desv_std = df["tiempo_entrega_horas"].std()
desv_std
```

### Coeficiente de Variación

```python
coef_var = (desv_std / media) * 100
coef_var
```
