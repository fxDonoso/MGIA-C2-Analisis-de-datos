```python
#Valores nulos
print(df.isnull().sum())
```

```python
df["tiempo_entrega_horas"] = pd.to_numeric(df["tiempo_entrega_horas"], errors="coerce")
```

```python
df
```

```python
media = df["tiempo_entrega_horas"].mean()
print("La media de tiempos de entrega en horas es de ", media)
```

```python
mediana = df["tiempo_entrega_horas"].median()
print("La mediana de tiempos de entrega es: ", mediana)
```

```python
moda = df["tiempo_entrega_horas"].mode()
print("La moda es: ", moda)
```

Medidas de dispersión

```python
varianza = df["tiempo_entrega_horas"].var()
varianza
```

```python
desv_std = df["tiempo_entrega_horas"].std()
desv_std
```

```python
moda = df["tiempo_entrega_horas"].mode()
moda
```

```python
coef_var = (desv_std / media ) * 100
coef_var
```

```python
df["tiempo_entrega_horas"].describe()
```

```python
resumen_zona = df.groupby("zona")["tiempo_entrega_horas"].agg(
    media="mean",
    mediana = "median",
    varianza = "var",
    desv_std = "std",
    minimo = "min",
    maximo = "max",
    conteo = "count"
)
```

```python
resumen_zona
```

```python
resumen_zona["coef_variacon_%"] = (
    resumen_zona["desv_std"] / resumen_zona["media"]
) * 100
```

```python
resumen_zona
```

```python
resumen_zona.sort_values(by="media", ascending=False)
```

```python
print(f"El tiempo promedio general de entregas es de {media:.2f} horas.")
print(f"La mediana es de {mediana:.2f} horas.")
```

```python
if media > mediana:
    print("La media es mayor que la mediana, lo que puede sugerir valores extremos altos")
elif media < mediana:
    print("La meddia es menor que la mediana, lo que puede sugerir valores extremos bajos")
else:
    print("La media y la mediana son muy similares")
```

```python
zona_mayor_media = resumen_zona["media"].idxmax()
zona_mayor_media
```

```python
zona_mayor_cv = resumen_zona["coef_variacon_%"].idxmax()
zona_mayor_cv
```

```python
print(f"La zona con mayor tiempo promedio es {zona_mayor_media}")
```

```python
print(f"La zona con mayor variabilidad relativa es {zona_mayor_cv}")
```
