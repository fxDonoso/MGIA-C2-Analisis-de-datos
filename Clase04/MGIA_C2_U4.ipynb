Homogeniza generos

```python
# Verificar valores únicos antes de limpiar
print("Antes:")
print(df3['victim_gender'].value_counts(dropna=False))

# Paso 1: normalizar a minúsculas y sin espacios
df3['victim_gender'] = df3['victim_gender'].str.strip().str.lower()

# Paso 2: mapear al valor canónico
mapa_genero = {
    'm':        'Male',
    'male':     'Male',
    'f':        'Female',
    'female':   'Female',
    'unknown':  'Unknown',
    'other':    'Other',
}

df3['victim_gender'] = df3['victim_gender'].map(mapa_genero)

# Verificar resultado
print("\nDespués:")
print(df3['victim_gender'].value_counts(dropna=False))
```
