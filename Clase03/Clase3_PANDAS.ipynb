**Trabajo con DataSet 311 Service Request**

```python
import requests, pandas as pd
```

```python
url = "https://data.cityofnewyork.us/resource/erm2-nwe9.json"
params = {"$limit": 10000}
r = requests.get(url, params = params, timeout=60)
```

```python
r.raise_for_status()
```

```python
df = pd.DataFrame(r.json())
#print(r.json())
```

```python
df.head(10)
```

```python
df.shape
```

```python
#Revisar algunas columnas importantes
df[["unique_key", "created_date", "agency_name"]].head()
```

```python
#Revisar cantidad de valores nulos por columna
df.isna().sum().sort_values(ascending=False)
```

```python
#Tipos de reclamos más frecuentes
df["complaint_type"].value_counts().head(10)
```

```python
df["borough"].value_counts()
```

```python
df_brooklyn = df[df["borough"] == "BROOKLYN"]
```

```python
df_brooklyn
```

```python
df_brooklyn.head()
```

```python
#filtrar por ruido
df_noise = df[df["complaint_type"].str.contains("Noise", na=False)]
```

```python
df_noise.head()
```

```python
import matplotlib.pyplot as plt

reclamos_distrito = df["borough"].value_counts()
```

```python
reclamos_distrito
```

```python
plt.figure(figsize=(8,5))
plt.bar(reclamos_distrito.index, reclamos_distrito.values)
plt.title("Cantidad de reclamos plataforma 311 Service Request por distrito")
plt.xlabel("Distrito")
plt.ylabel("Cantidad de reclamos")
plt.xticks(rotation=45)
plt.show()
```

```python
top_reclamos = df["complaint_type"].value_counts().head(10)
```

```python
top_reclamos
```

```python
plt.figure(figsize=(10,6))
plt.barh(top_reclamos.index, top_reclamos.values)
plt.title("Top 10 tipos de reclamos 311 Service Request")
plt.xlabel("Cantidad de reclamos")
plt.ylabel("Tipo de reclamo")
plt.show()
```
